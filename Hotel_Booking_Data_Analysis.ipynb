# Hotel Booking Data Analysis

Bu layihədə Python və Pandas vasitəsilə 119,390 otel rezervasiyasından ibarət dataset üzərində məlumatların təmizlənməsi, yeni göstəricilərin yaradılması və exploratory data analysis aparmışam.

Analizdə rezervasiyaların ləğv statusu, lead time, ADR, otel növü, müştəri tipi, gecələmə müddəti və hesablanmış rezervasiya dəyəri arasındakı əlaqələri araşdırmışam.


```python
import pandas as pd
```

```python
df = pd.read_csv(r"Data/Hotel_Bookings.csv")
```

```python
# EDA - Exploratory Data Analysis
# Sətir və sütunların sayını göstərir.
df.shape
```

```python
# Data Cleaning
# Bütün sütunları null (NaN) olan, yəni boş olan sətirləri silirik. 
df.dropna(how="all", inplace=True)
```

```python
# Data Cleaning
# Datada olan sütunlardakı null sətirlərin faiz nisbəti
(df.isnull().sum()/len(df))*100
```

```python
# Data Cleaning
# Datada olan sütunlardakı null sətirlərin cəmi
df.isnull().sum()
```

```python
# EDA - Exploratory Data Analysis
# dtype-lafrını göstərir
df.dtypes
```

```python
# EDA - Exploratory Data Analysis
# Datanın hansı sətirlərində nə qədər null olub olmadığını göstərir.
df.info()
```

```python
# EDA - Exploratory Data Analysis
# Sütunların adlarını göstərir.
df.columns
```

```python
# Data Cleaning
# "reservation_status_date" sütunu tarixdir,lakin object kimi qeyd olunub, onu tarixə çevirmək üçün bunu edirik:
df["reservation_status_date"] = pd.to_datetime(df["reservation_status_date"], errors="coerce")
```

```python
# EDA - Exploratory Data Analysis
# Datanın hansı sətirlərində nə qədər null olub olmadığını təzədən yoxlayırıq
df.info()
```

```python
# EDA - Exploratory Data Analysis
# DataFrame-dəki ədədi (numeric) sütunlar haqqında ümumi statistik məlumat.
df.describe()
```

```python
# EDA - Exploratory Data Analysis
df.tail(1)
```

```python
# Data Cleaning
# Cədvəldə olan 3 sütun üzrə arrival_date_year , arrival_date_month , arrival_date_day_of_month vahid sütun yaradılır.
df["arrival_date"] = pd.to_datetime(
    df["arrival_date_year"].astype(str) + "-" +
    df["arrival_date_month"] + "-" +
    df["arrival_date_day_of_month"].astype(str),
    format="%Y-%B-%d"
)
```

```python
# EDA - Exploratory Data Analysis
df.head(1)
```

```python
# EDA - Exploratory Data Analysis
# Sütundakı fərqli dəyərlərin sayı
df.nunique()
```

```python
# Data Cleaning
# Dublikatların yoxlanması
df.duplicated().sum()
```

```python
# Feature Engineering
# Ümumi gecə sayı
df["total_nights"] = (
    df["stays_in_week_nights"] +
    df["stays_in_weekend_nights"]
)

# EDA - Exploratory Data Analysis
df.tail(1)
```

```python
# Feature Engineering
# Ümumi qonaq sayı
df["total_guests"] = (
    df["adults"] +
    df["children"] +
    df["babies"]
)

# EDA - Exploratory Data Analysis
df.head(1)
```

```python
# Feature Engineering
# Reservation ayı
df["month"] = df["arrival_date"].dt.month

# EDA - Exploratory Data Analysis
df.head(1)
```

```python
# Feature Engineering
# Reservation həftəsi
df["weekday"] = df["arrival_date"].dt.day_name()

# EDA - Exploratory Data Analysis
df.head(1)
```

```python
# Insight
# Rezervasiya olunan otellərdəki miqdarlar.Ən çox rezervasiya olunan otel artıq tapılmışdır.
df["hotel"].value_counts()
```

```python
# Insight
# Ləğv olunma və ya Cancellation faizi.
df["is_canceled"].value_counts(normalize=True)*100
```

```python
# Insight
# Top 10 Ölkələr üzrə rezervasiya olunma miqdarı
df["country"].value_counts().head(10)
```

```python
# Insight
# Ən çox rezervasiya olunan aylar
df["arrival_date_month"].value_counts()
```

```python
# Insight
# ADR ortalaması - ADR (Average Daily Rate) — Otelin bir gecə üçün orta satış qiymətidir.
df.groupby("hotel")["adr"].mean()
```

```python
# EDA - Exploratory Data Analysis
df.head(1)
```

```python
df["total_price"] = df["total_nights"] * df["adr"]
df.head()
```

```python
# profit = df[df["is_canceled"]==0].groupby("arrival_date_month")["adr"]
# Ləğv olunmayan rezervasiyalarda hər ay neçə rezervasiya olub?
df[df["is_canceled"]==0].groupby("arrival_date_month").size().sort_values()
```

```python
# Təsdiqlənmiş rezervasiyalardan ən çox gəlir gətirən ay hansıdır?
df[df["is_canceled"] == 0].groupby("arrival_date_month")["total_price"].sum().sort_values()
```

```python
# Ən çox təsdiq olunmuş əməliyyatı olan market seqmenti
df[df["is_canceled"] == 0].groupby("market_segment").count()
```

```python
# Müştəri tipinə görə rezervasiya sayı
df.groupby("customer_type")["total_nights"].count()
```

```python
# Müştəri tipinə görə ümumi gecələmə sayı
df.groupby("customer_type")["total_nights"].sum()
```

```python
# Müştəri tipinə görə orta gecələmə müddəti
df.groupby("customer_type")["total_nights"].sum() / df.groupby("customer_type")["total_nights"].count()
```

```python
# Ləğv statusuna görə rezervasiya sayı
df.groupby("is_canceled")["lead_time"].count()
```

```python
# Ləğv statusunun faizlə bölgüsü
df["is_canceled"].value_counts(normalize=True) * 100
```

```python
# Ləğv statusuna görə orta lead time
df.groupby("is_canceled")["lead_time"].mean()
```

## Yekun nəticə

Bu layihədə 119,390 otel rezervasiyası üzrə məlumatları yoxladım, tarix sütunlarını emal etdim və analiz üçün yeni göstəricilər yaratdım.

* Rezervasiyaların 37.04%-i ləğv edilib.
* City Hotel 79,330 rezervasiya ilə Resort Hotel-dən daha çox rezervasiya olunub.
* Ləğv edilmiş rezervasiyalarda orta lead time 144.85 gün, ləğv edilməyən rezervasiyalarda isə 79.98 gün olub. Bu nəticə daha erkən edilən rezervasiyalarda ləğv səviyyəsinin daha yüksək ola biləcəyini göstərir, lakin birbaşa səbəb əlaqəsini sübut etmir.
* City Hotel üzrə orta ADR 105.30, Resort Hotel üzrə isə 94.95 olub.
* Təsdiqlənmiş rezervasiyaların sayı və hesablanmış rezervasiya dəyəri ən yüksək avqust ayında qeydə alınıb.
* Contract müştəri tipində orta gecələmə müddəti 5.32 gecə ilə digər müştəri tiplərindən daha yüksək olub.

## Məlumatların emalı ilə bağlı qeyd

Yalnız bütün sütunları boş olan sətirləri silmək üçün `dropna(how="all")` yanaşmasından istifadə etdim. `agent` və `company` sütunlarındakı boş dəyərləri silmədim, çünki həmin sətirlərdə rezervasiyaya aid digər məlumatlar mənalıdır. Bu boşluqlar rezervasiyanın agent və ya şirkət vasitəsilə edilmədiyini göstərə bilər; buna görə sətirlərin tam silinməsi analizin nəticələrini təhrif edə bilərdi.

`total_price` göstəricisini ADR ilə ümumi gecələmə sayının hasili kimi hesabladım. Buna görə bu göstərici rəsmi gəlir deyil, dataset əsasında hesablanmış təxmini rezervasiya dəyəridir.

Dataset təlim və analiz məqsədi daşıyır. Nəticələr real otel şirkətinin cari fəaliyyət göstəriciləri kimi qəbul edilməməlidir.
